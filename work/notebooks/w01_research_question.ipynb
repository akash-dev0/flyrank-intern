```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Number 1: CTR by position tier — does position actually drive clicks?
visible = df[df["impressions_90d"] >= 100]
ctr_by_pos = visible.groupby("position_tier")["ctr"].mean().sort_values(ascending=False)
print("CTR by position tier:")
print(ctr_by_pos.round(4).to_string())

# Number 2: does keyword search volume predict actual traffic?
corr = df["search_volume"].corr(df["impressions_90d"])
print(f"\nCorrelation(search_volume, impressions_90d): {corr:.3f}")

# Number 3: does word count separate declining pages from growing pages?
wc = df.groupby("trend_direction")["word_count"].median()
print(f"\nMedian word_count by trend_direction:")
print(wc.round(0).to_string())
```

**What I cannot claim:** causation. I can't say a page moves up in rankings

observing a snapshot. I also can't claim to have reverse-engineered Google's
ranking algorithm, or that any correlation I find will hold for a client outside
