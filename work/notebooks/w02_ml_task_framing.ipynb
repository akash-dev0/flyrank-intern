## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

The target is `is_declining_label`, which the prep pipeline sets to 1 when
`trend_direction == "down"`, and 0 otherwise.

This is a **defined rule, not an observed future outcome**. `trend_direction`
itself is calculated by comparing impressions in the last 30 days to the
previous 30 days — a drop of more than 20% is labeled "down." So the label is
a threshold applied to current-window data, not a verified real-world result
like "this page's traffic kept falling for the next quarter" or "the client
lost revenue."

Because of this, `trend_direction` and `trend_pct` must never be used as
model features — they are the exact source of the label, so including them
would leak the answer straight into the input

## 3. Success metric

*One metric you can defend. What number means 'good'?*

The metric that matches the real decision is **Precision@50**: of the top 50
pages the model ranks highest for review, how many are actually declining
(by our proxy label)?

This fits the real workflow better than plain accuracy, because the output
isn't "classify every page correctly" — it's a ranked queue that a reviewer
works through with limited capacity, starting from the top. Precision@K
measures exactly that: how good the top of the list actually is.

As a reference point, the starter pipeline's own results show why this
matters: a fixed baseline rule scores Precision@50 = 0.240 (about 12 of the
top 50 are correct), while a random forest model scores Precision@50 = 0.740
(about 37 of the top 50 are correct). That gap is the evidence that a learned
model can meaningfully beat a hand-written rule on this data.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule like `stale_visible_page` (`days_since_last_update >= 180 and
impressions_90d >= 500`) only checks one or two columns against a hard
cutoff. But whether a page is actually at risk of declining depends on many
signals interacting at once — word count, CTR, position, engagement rate,
scroll rate, content age, competition level, intent — and no single
threshold can capture how these combine. A page might pass every individual
rule and still be declining because of a combination none of the rules
account for.

This is exactly what the starter pipeline's own results show: the fixed
baseline rules scored Precision@50 = 0.240, while the random forest —
which can weigh dozens of features together and learn nonlinear
combinations — scored 0.740. That gap is the evidence: with 44 interacting
columns, a model finds patterns a human-written rule set misses.
